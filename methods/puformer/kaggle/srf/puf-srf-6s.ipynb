```python
import subprocess, sys, json, numpy as np
subprocess.run([sys.executable, "-m", "pip", "install", "-q", "Py6S"], check=False)
from Py6S import PredefinedWavelengths as PW
out = {}
for name in dir(PW):
    if name.startswith("_"):
        continue
    v = getattr(PW, name)
    if isinstance(v, tuple) and len(v) == 4 and hasattr(v[3], "__len__") and len(v[3]) > 1:
        out[name] = dict(id=int(v[0]), start_um=float(v[1]), end_um=float(v[2]), values=[float(x) for x in v[3]])
json.dump(out, open("/kaggle/working/srf_6s.json", "w"))
print(len(out), "filter functions;", sorted(k for k in out if any(t in k for t in ("WORLDVIEW", "QUICKBIRD", "IKONOS", "OLI", "S2A", "S2B", "GEOEYE", "PLANET")))[:80])
```
