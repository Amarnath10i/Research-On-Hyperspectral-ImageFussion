# Diagnose Chikusei dataset path on Kaggle

```python
import os, glob
print('=== All files in /kaggle/input ===')
for root, dirs, files in os.walk('/kaggle/input'):
    level = root.replace('/kaggle/input', '').count(os.sep)
    indent = '  ' * level
    basename = os.path.basename(root)
    print(f'{indent}{basename}/')
    for f in files:
        fp = os.path.join(root, f)
        sz = os.path.getsize(fp) / 1e6
        print(f'{indent}  {f} ({sz:.1f} MB)')
    if level > 3:
        dirs[:] = []
print()
print('=== .mat files > 1MB ===')
for m in glob.glob('/kaggle/input/**/*.mat', recursive=True):
    print(f'  {m} ({os.path.getsize(m)/1e6:.1f} MB)')
```
